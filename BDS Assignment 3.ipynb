### Task 0: DGP

```python
# Packages inmported
import pandas as pd
import statsmodels.api as sm
import numpy as np

np.random.seed(2026)
```

```python
def dgp(n=500):
    # beta
    beta0 = 1
    beta = np.zeros(20)
    beta[:10] = 0.1 * np.arange(1, 11) 

    # error
    eps = np.random.normal(0, 4, size=n)
      
    # Regressors
    ## (X1, . . . , X5) and (X11, . . . , X15) have a standard multivariate normal
    X_blok1 = np.random.normal(0, 1, size=(n, 5))
    X_blok3 = np.random.normal(0, 1, size=(n, 5))
    
    ## (X6, . . . , X10) and (X16, . . . , X20) have multivariate normal
    cov = np.array([
        [4, 1, 0, 0, 0],
        [1, 4, 1, 0, 0],
        [0, 1, 6, 2, 0],
        [0, 0, 2, 4, 2],
        [0, 0, 0, 2, 4]
    ])
    
    mean = np.ones(5) * 2
    
    X_blok2 = np.random.multivariate_normal(mean, cov, size=n)
    X_blok4 = np.random.multivariate_normal(mean, cov, size=n)

    ## Combine all blocks
    X = np.hstack([X_blok1, X_blok2, X_blok3, X_blok4])

    # Y
    
    Y = beta0 + X @ beta + eps

    return X, Y
```

### Task 5: OLS

```python
def task1(X, Y, alpha=0.05):
    
    Xt = sm.add_constant(X)
    model = sm.OLS(Y, X).fit()
    
    pvals = model.pvalues[1:]  # must contain the intercept?
    
    selected = set(np.where(pvals < alpha)[0])
    
    return selected
```

```python
def monte_carlo_task5(M=1000):
    
    correct = 0
    true_set = set(range(10))
    
    for m in range(M):
        X, Y = dgp()
        
        selected = task1(X, Y)
        
        if selected == true_set:
            correct += 1
    
    return correct / M, true_set, selected
```
