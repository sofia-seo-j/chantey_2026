### Set up

```python
import numpy as np
import random
random.seed(0)
```

```python
T = 5000
n_machines = 10

# true probabilities
p_true = []
for i in range(1, 11):
    if i <= 5:
        p_true.append(0.12 * i)
    else:
        p_true.append(1 - (i - 1) / 10)

# initial guesses
p0 = [0.7 - i*0.05 for i in range(10)]
p_hat = p0.copy()

plays = [0]*n_machines
successes = [0]*n_machines

total_reward = 0
success_first100 = 0
p1_hat_round2 = None

for t in range(1, T+1):

    if t == 2:
        p1_hat_round2 = p_hat[0]

    # choose machine
    if t % 10 == 0:
        machine = random.randint(0,9)
    else:
        machine = p_hat.index(max(p_hat))

    # simulate play
    win = random.random() < p_true[machine]

    plays[machine] += 1

    if win:
        successes[machine] += 1
        total_reward += 1

    if t <= 100 and win:
        success_first100 += 1

    # update estimates
    for i in range(n_machines):
        p_hat[i] = (
            p0[i] / (1 + plays[i]) +
            successes[i] / (1 + plays[i])
        )

print("\nSlot Machine Results\n")
print(f"{'Machine':<10}{'True p':<10}{'Estimated p':<15}{'Plays':<10}{'Successes':<12}{'Empirical p':<12}")
print("-"*70)

for i in range(n_machines):
    empirical = successes[i]/plays[i] if plays[i] > 0 else 0
    print(f"M{i+1:<9}{p_true[i]:<10.3f}{p_hat[i]:<15.3f}{plays[i]:<10}{successes[i]:<12}{empirical:<12.3f}")

print("-"*70)
print(f"{'Total reward':<20}: {total_reward}")
print(f"{'Total plays':<20}: {sum(plays)}")
print(f"{'Avg success':<20}: {total_reward/sum(plays):.3f}")

print("\nAnswers for quiz:")
print("Q1:", round(p1_hat_round2,3))
print("Q2:", round(success_first100/100,3))
print("Q3:", round(total_reward/5000,3))
```
