Below you find a randomly initialized deep neural network.

**Your Task**
- Implement the *relu(z)* function: $\text{ReLU}(z) = \max(0, z)$.

- Implement the *forward_pass(x, params)* function. It should iterate through the layers performing the operation
$$\mathbf{h} = \text{ReLU}(\mathbf{W}\mathbf{x} + \mathbf{b}),$$
where the last layer does not apply the ReLU activation function.

- Calculate the output $y$ for the specific input $x = 0.75$.

```python
import numpy as np

# 1. Configuration
np.random.seed(42)

# Architecture: Input(1) -> 10 -> 10 -> 10 -> 10 -> Output(1)
layer_sizes = [1, 10, 10, 10, 10, 1]

# 2. Parameter Initialization
parameters = []

print("Network Architecture & Shapes:")
for i in range(len(layer_sizes) - 1):
    n_in = layer_sizes[i]
    n_out = layer_sizes[i+1]

    # He Initialization: Scale weights by sqrt(2/n_in)
    std_dev = np.sqrt(2 / n_in)
    W = np.random.randn(n_out, n_in) * std_dev

    # Initialize biases to small positive value to prevent "Dying ReLU"
    b = np.ones((n_out, 1)) * 0.01

    parameters.append((W, b))
    print(f"Layer {i+1}: W shape {W.shape}, b shape {b.shape}")

# 3. The Input
# x is a column vector of shape (1, 1)
x_input = np.array([[0.75]])

print(f"\nInput x:\n{x_input}")
```

```python
# 4. Implement ReLU
def relu(z):
    return np.maximum(0, z)

# 5. Implement forward pass
def forward_pass(x, params):
    h = x
    for idx, (W, b) in enumerate(params):
        z = W @ h + b
        if idx < len(params) - 1:
            h = relu(z)     # hidden layers
        else:
            h = z           # output layer (linear)
    return h

# 6. Compute output
y = forward_pass(x_input, parameters)
print("\nNetwork output y:")
print(y)
```
